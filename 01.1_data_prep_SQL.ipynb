```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("pricing_project.db")
```

```python
df_freq = pd.read_csv("freMTPL2freq.csv")
df_sev = pd.read_csv("freMTPL2sev.csv")

df_freq.to_sql("freq", conn, if_exists='replace', index=False)
df_sev.to_sql("sev", conn, if_exists='replace', index=False)
```

```python
full_pop_df = pd.read_sql("SELECT * FROM cleaned_policies", conn)
full_pop_df.describe()
```

```python
claim_df = pd.read_sql("SELECT * FROM claim_only", conn)
claim_df.describe()
```

```python
full_pop_train = full_pop_df.sample(frac=0.8, random_state=1)
full_pop_test = full_pop_df.drop(full_pop_train.index)

claim_train = full_pop_train[full_pop_train["ClaimTotal"] > 0].copy()
claim_test = full_pop_test[full_pop_test["ClaimTotal"] > 0].copy()
```
