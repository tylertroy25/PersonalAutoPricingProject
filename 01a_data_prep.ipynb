```python
import pandas as pd

df_sev = pd.read_csv("freMTPL2sev.csv")
df_sev["IDpol"] = df_sev["IDpol"].astype("int64")
df_freq = pd.read_csv("freMTPL2freq.csv")
df_freq["IDpol"] = df_freq["IDpol"].astype("int64")
```

```python
data = {
    'IDpol': [],
    'ClaimNb': [],
    'ClaimTotal': [],
    'ClaimTotalCapped': [],
    'Exposure': [],
    'Area': [],
    'VehPower': [],
    'VehAge': [],
    'DrivAge': [],
    'BonusMalus': [],
    'VehBrand': [],
    'VehGas': [],
    'Density': [],
    'Region': [],
}
```

```python
for col, row in df_freq.iterrows():
    ID = row["IDpol"]

    data["IDpol"].append(ID)
    data["ClaimNb"].append(row["ClaimNb"])
    data["ClaimTotal"].append(df_sev.loc[df_sev["IDpol"] == ID, "ClaimAmount"].sum())
    data["ClaimTotalCapped"].append(df_sev.loc[df_sev["IDpol"] == ID, "ClaimAmount"].clip(upper=25000).sum())
    data["Exposure"].append(row["Exposure"])
    data["Area"].append(row["Area"])
    data["VehPower"].append(row["VehPower"])
    data["VehAge"].append(row["VehAge"])
    data["DrivAge"].append(row["DrivAge"])
    data["BonusMalus"].append(row["BonusMalus"])
    data["VehBrand"].append(row["VehBrand"])
    data["VehGas"].append(row["VehGas"])
    data["Density"].append(row["Density"])
    data["Region"].append(row["Region"])
```

```python
full_pop_df = pd.DataFrame(data)
full_pop_df["Exposure"] = full_pop_df["Exposure"].clip(upper=1.0)
full_pop_df["ClaimNb"] = full_pop_df["ClaimNb"].clip(upper=11.0)
full_pop_df.describe()
```

```python
claim_df = full_pop_df[full_pop_df["ClaimTotal"] > 0]
claim_df.describe()
```

```python
full_pop_train = full_pop_df.sample(frac=0.8, random_state=1)
full_pop_test = full_pop_df.drop(full_pop_train.index)

claim_train = full_pop_train[full_pop_train["ClaimTotal"] > 0].copy()
claim_test = full_pop_test[full_pop_test["ClaimTotal"] > 0].copy()
```
