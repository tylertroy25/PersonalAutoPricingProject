```python
import matplotlib.pyplot as plt

%store -r freq_validation
%store -r sev_validation
```

```python
plt.plot(freq_validation.index, freq_validation["actual_freq"], label="Actual")
plt.plot(freq_validation.index, freq_validation["predicted_freq"], label="Predicted")
plt.xlabel("Decile")
plt.ylabel("Frequency")
plt.legend()
plt.title("Actual vs Predicted Frequency by Decile")
plt.show()
```

```python
plt.plot(sev_validation.index, sev_validation["actual_sev"], label="Actual")
plt.plot(sev_validation.index, sev_validation["predicted_sev"], label="Predicted")
plt.xlabel("Decile")
plt.ylabel("Severity")
plt.legend()
plt.title("Actual vs Predicted Severity by Decile")
plt.show()
```
